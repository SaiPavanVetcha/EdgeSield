```python
%sql
USE SCHEMA bronze;

CREATE TABLE IF NOT EXISTS environment_metrics (
    event_id STRING NOT NULL,
    device_id STRING NOT NULL,
    timestamp TIMESTAMP,
    temperature_c DOUBLE,
    humidity_percent DOUBLE,
    _ingested_at TIMESTAMP
) USING DELTA
TBLPROPERTIES (
    'delta.enableChangeDataFeed' = 'true',
    'delta.autoOptimize.optimizeWrite' = 'true'
);

CREATE TABLE IF NOT EXISTS system_metrics (
    event_id STRING NOT NULL,
    device_id STRING NOT NULL,
    hostname STRING,
    timestamp TIMESTAMP,
    platform STRING,
    boot_time_utc TIMESTAMP,
    _ingested_at TIMESTAMP
) USING DELTA
TBLPROPERTIES (
    'delta.enableChangeDataFeed' = 'true',
    'delta.autoOptimize.optimizeWrite' = 'true'
);

CREATE TABLE IF NOT EXISTS cpu_metrics (
    event_id STRING NOT NULL,
    device_id STRING NOT NULL,
    timestamp TIMESTAMP,
    utilization_percent DOUBLE,
    logical_cores INT,
    physical_cores INT,
    frequency_current_mhz DOUBLE,
    frequency_max_mhz DOUBLE,
    _ingested_at TIMESTAMP
) USING DELTA
TBLPROPERTIES (
    'delta.enableChangeDataFeed' = 'true',
    'delta.autoOptimize.optimizeWrite' = 'true'
);

CREATE TABLE IF NOT EXISTS ram_metrics (
    event_id STRING NOT NULL,
    device_id STRING NOT NULL,
    timestamp TIMESTAMP,
    total_bytes BIGINT,
    used_bytes BIGINT,
    available_bytes BIGINT,
    utilization_percent DOUBLE,
    available_mb DOUBLE,
    _ingested_at TIMESTAMP
) USING DELTA
TBLPROPERTIES (
    'delta.enableChangeDataFeed' = 'true',
    'delta.autoOptimize.optimizeWrite' = 'true'
);

CREATE TABLE IF NOT EXISTS disk_metrics (
    event_id STRING NOT NULL,
    device_id STRING NOT NULL,
    timestamp TIMESTAMP,
    mount_point STRING,
    total_bytes BIGINT,
    used_bytes BIGINT,
    free_bytes BIGINT,
    free_gb DOUBLE,
    utilization_percent DOUBLE,
    read_bytes BIGINT,
    write_bytes BIGINT,
    _ingested_at TIMESTAMP
) USING DELTA
TBLPROPERTIES (
    'delta.enableChangeDataFeed' = 'true',
    'delta.autoOptimize.optimizeWrite' = 'true'
);

CREATE TABLE IF NOT EXISTS battery_metrics (
    event_id STRING NOT NULL,
    device_id STRING NOT NULL,
    timestamp TIMESTAMP,
    present BOOLEAN,
    percent DOUBLE,
    power_plugged BOOLEAN,
    design_capacity_raw DOUBLE,
    full_charge_capacity_raw DOUBLE,
    health_percentage DOUBLE,
    _ingested_at TIMESTAMP
) USING DELTA
TBLPROPERTIES (
    'delta.enableChangeDataFeed' = 'true',
    'delta.autoOptimize.optimizeWrite' = 'true'
);
```
