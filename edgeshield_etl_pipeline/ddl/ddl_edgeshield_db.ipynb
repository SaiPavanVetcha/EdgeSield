```python
%sql
CREATE DATABASE IF NOT EXISTS edgeshield_db;
USE edgeshield_db;
```

```python
%sql
-- 1. System Metrics Table
CREATE TABLE IF NOT EXISTS system_metrics (
    event_id STRING,
    device_id STRING,
    hostname STRING,
    timestamp TIMESTAMP,
    platform STRING,
    boot_time_utc TIMESTAMP
) USING DELTA;

-- 2. CPU Metrics Table
CREATE TABLE IF NOT EXISTS cpu_metrics (
    event_id STRING,
    device_id STRING,
    timestamp TIMESTAMP,
    utilization_percent DOUBLE,
    logical_cores INT,
    physical_cores INT,
    frequency_current_mhz DOUBLE,
    frequency_max_mhz DOUBLE
) USING DELTA;

-- 3. RAM Metrics Table
CREATE TABLE IF NOT EXISTS ram_metrics (
    event_id STRING,
    device_id STRING,
    timestamp TIMESTAMP,
    total_bytes BIGINT,
    used_bytes BIGINT,
    available_bytes BIGINT,
    utilization_percent DOUBLE,
    available_mb DOUBLE
) USING DELTA;

-- 4. Disk Metrics Table
CREATE TABLE IF NOT EXISTS disk_metrics (
    event_id STRING,
    device_id STRING,
    timestamp TIMESTAMP,
    mount_point STRING,
    total_bytes BIGINT,
    used_bytes BIGINT,
    free_bytes BIGINT,
    free_gb DOUBLE,
    utilization_percent DOUBLE,
    read_bytes BIGINT,
    write_bytes BIGINT
) USING DELTA;

-- 5. Battery Metrics Table
CREATE TABLE IF NOT EXISTS battery_metrics (
    event_id STRING,
    device_id STRING,
    timestamp TIMESTAMP,
    present BOOLEAN,
    percent DOUBLE,
    power_plugged BOOLEAN,
    design_capacity_raw BIGINT,
    full_charge_capacity_raw BIGINT,
    health_percentage DOUBLE
) USING DELTA;

-- 6. Environment Metrics Table
CREATE TABLE IF NOT EXISTS edgeshield_db.environment_metrics (
    event_id STRING,
    device_id STRING,
    timestamp TIMESTAMP,
    temperature_c DOUBLE,
    humidity_percent DOUBLE
) USING DELTA;
```
